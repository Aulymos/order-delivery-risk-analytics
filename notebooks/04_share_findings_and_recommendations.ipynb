# 04 Share - Delivery Risk Findings and Review Recommendation

## 1. Define the Share scope and evident handoff

* **Audience:** Operations and delivery decision-makers responsible for allocating limited review capacity.  
  
* **Decision:** Determine whether a targeted review process should be introduced and which orders should be reviewed first.  
  
* **Communication focus:** Present the AQ1–AQ4 findings and connect the evidence to a controlled top-10% review pilot.  
  
* **Communication boundary:** Use the validated Analyze-stage results without repeating calculations or presenting the risk-ranking model as a confirmed production solution.


## 2. Explain AQ1–AQ4 into communication messages

### 2.1 AQ1 — Establish the scale and persistence of late delivery

* **Measure and context:** `Late-delivery rate` equals `late orders` **divided** by `eligible orders`. The historical scope contains ``one eligible order per row`` from January 2015 to January 2018.  
  
* **Selected visual:** Three **KPI cards** present the overall measures, followed by **a monthly trend chart** with the `overall late-delivery rate` as a **reference line**. The monthly view was selected to show the longer-term pattern clearly.  
  
* **Main insight:** Late delivery remained **high** throughout the historical period, with no sustained upward or downward trend.  
  
* **Supporting evidence:** Among **62,897** eligible orders, **36,048** were late, producing an ``overall late-delivery rate`` of **57.31%**. ``Monthly rates`` remained between **55.04%** and **59.86%**, while ``monthly order volumes`` were broadly stable.  
  
* **Business meaning:** Late delivery appears to be a **persistent issue** rather than a temporary spike, indicating the requirement of an **ongoing prioritisation process** instead of a one-off response.

### 2.2 AQ2 — Explain the clearest operational association

* **Measure and context:** ``Segment late-delivery rate`` equals ``late orders`` divided by ``eligible orders within each category``. The comparison uses attributes known at **order creation** and considers **exposure counts** alongside the rates.  
  
* **Selected visual:** A **bar chart** compares ``late-delivery rates`` across the **four Shipping Mode** categories. Two callouts contrast the Shipping Mode gap with the much smaller variation across Customer Segment and Market.  
  
* **Main insight:** **Shipping Mode** showed the **clearest observed association** with late delivery. **Customer Segment**, **Market**, and the selected **order-size measures** showed much weaker differences.  
  
* **Supporting evidence:** **First Class** recorded **100.00%** late delivery (9,602 of 9,602 orders), followed by **Second Class** at **79.99%**, **Same Day** at **48.37%**, and **Standard Class** at **39.85%**. The **First Class–Standard Class** gap was **60.15%**, while that across **Customer Segment** and **Market** was no more than **1.19%**.  
  
* **Business meaning:** **Shipping Mode** should be investigated and retained as a **risk-ranking signal**, but the observed association should not be interpreted as evidence that Shipping Mode causes late delivery.

### 2.3 AQ3 — Explain risk-ranking performance and its limits

* **Measure and context:** **Model performance** is evaluated on **13,525** orders from the later **32-week** held-out period. **Average Precision** and **ROC-AUC** compare a standard **logistic-regression** ranking with a **non-informative baseline**.  
  
* **Selected visual:** Metric panels explain **Average Precision**, **ROC-AUC**, and the **baseline**, followed by **KPI comparisons** and a **precision–recall curve** showing performance across different risk-score thresholds.  
  
* **Main insight:** **Logistic regression** ranked late orders **more effectively** than the **non-informative baseline** on the held-out test period.  
  
* **Supporting evidence:** **Average Precision** increased from **0.5759** to **0.8382**, while **ROC-AUC** increased from **0.5000** to **0.7473**. The **precision–recall** curve of **Logistic Regression model** remained **above** the **prevalence baseline** across almost the entire recall range.  
  
* **Business meaning:** The model scores provide useful relative prioritisation for the capacity simulation, but they do not guarantee individual delivery outcomes, establish the best possible model, or confirm performance outside the single held-out test period.

### 2.4 AQ4 — Explain the capacity trade-off and review queue

* **Measure and context:** The retrospective simulation applies **5%**, **10%**, and **20%** review-capacity limits to the **fixed ranking** of **13,525** held-out test orders. Each scenario is evaluated using ``late-order recall``, ``review precision``, and ``lift`` over the overall test-period late-delivery rate.  
  
* **Selected visual:** **Four metric panels** define the capacity measures, followed by a **scenario comparison chart** and a **summary** of the preferred top-10% queue.  
  
* **Main insight:** Increasing **review capacity** captured **more late orders** but **increased workload** and eventually **reduced review precision**. The top-10% queue provided a practical balance for a controlled pilot.  
  
* **Supporting evidence:** The 5% scenario reviewed 677 orders with 8.69% recall and 100.00% precision. The 10% scenario reviewed 1,353 orders with 17.37% recall, 100.00% precision, and 1.74x lift. At 20%, recall increased to 33.02%, while precision decreased to 95.08% and lift to 1.65x.  
  
* **Business meaning:** **Eligible orders** should be reviewed from the **highest risk rank** downward until the selected capacity limit is reached. The **top-10% rule** is a retrospective decision-support recommendation that requires prospective validation and does not show that reviewing an order will prevent late delivery.

## 3. Build the presentation story

### 3.1 Define the storytelling sequence

* **Decision context:** Introduce the persistent late-delivery problem, limited review capacity, and the need to prioritise orders for review.  
  
* **Problem scale — AQ1:** Establish how common late delivery was and whether it persisted over time.  
  
* **Operational association — AQ2:** Identify Shipping Mode as the clearest observed association while avoiding causal claims.  
  
* **Ranking evidence — AQ3:** Show that the logistic-regression model provided useful relative risk ranking on the temporal holdout, while clearly stating its limits.  
  
* **Capacity application — AQ4:** Translate the fixed risk ranking into 5%, 10%, and 20% review scenarios and compare workload with late-order capture.  
  
* **Recommendation:** Conclude with a controlled top-10% review pilot, supported by monitoring, prospective validation, and explicit limitations.  
  
**Story flow:** Decision context → problem scale → operational association → ranking evidence → capacity trade-off → controlled action.

### 3.2 Map the message flow to the 13 slides

* **Slide 1 — Title:** Introduce the portfolio case study and the task of prioritising late-delivery reviews under limited capacity.  
  
* **Slide 2 — Decision context:** Define the review-capacity problem, introduce AQ1–AQ4, and establish that the workflow supports prioritisation rather than automated decisions.  
  
* **Slide 3 — Data and method:** Summarise the historical scope, the standard logistic-regression model with a temporal holdout, and the later-period capacity simulation.  
  
* **Slide 4 — AQ1 measure:** Define the late-delivery rate and present the overall eligible-order, late-order, and rate measures.  
  
* **Slide 5 — AQ1 evidence:** Show that late delivery remained high throughout the historical period without a sustained trend.  
  
* **Slide 6 — AQ2 measure:** Define the segment comparison measures and retain exposure counts and the non-causal interpretation boundary.  
  
* **Slide 7 — AQ2 evidence:** Present Shipping Mode as the clearest observed association with late delivery.  
  
* **Slide 8 — AQ3 measure:** Explain Average Precision, ROC-AUC, and their non-informative baselines on the held-out test period.  
  
* **Slide 9 — AQ3 evidence:** Show that logistic regression ranked late orders better than baseline while framing the score as relative prioritisation.  
  
* **Slide 10 — AQ4 measure:** Define review capacity, late-order recall, review precision, lift, and the highest-risk-first decision rule.  
  
* **Slide 11 — AQ4 evidence:** Compare the 5%, 10%, and 20% scenarios and show the trade-off between late-order coverage and review efficiency.  
  
* **Slide 12 — Recommendation:** Present the top-10% queue as the preferred controlled pilot and show illustrative top-ranked orders.  
  
* **Slide 13 — Action and limitations:** Define the pilot, monitoring, and validation steps while carrying forward the main analytical limitations.

### 3.3 Separate slide, transcript, and technical-detail roles

* **Slide role:** Present one decision-relevant message per slide using concise audience-facing text, selected visuals, and only the essential numbers.  
  
* **Transcript role:** Provide the spoken explanation, including measure definitions, transitions, interpretation, business meaning, and necessary limitations without placing all details on the slides.  
  
* **Technical-detail role:** Retain calculations, model setup, evaluation design, supporting tables, and figures in the Analyze Notebook and `reports/analyze/` outputs for verification and follow-up questions.  
  
* **Alignment rule:** Every slide and transcript claim must remain consistent with the validated Analyze evidence.  
  
* **Separation principle:** Slides show what the audience needs to see, the transcript records what the presenter needs to explain, and the technical artifacts show how the evidence was produced.

## 5. State the recommendation, actions, and limitations


### 5.1 Recommend a controlled top-10% pilot

* **Recommendation:** Introduce a controlled pilot that reviews the top 10% of eligible orders ranked by predicted late-delivery risk.  
  
* **Selection rule:** Rank eligible orders from highest to lowest risk score and review orders until the 10% capacity limit is reached.  
  
* **Evidence basis:** In the held-out-period simulation, the top-10% scenario reviewed 1,353 of 13,525 orders and captured 17.37% of late orders, with 100.00% review precision and 1.74x lift.  
  
* **Rationale:** The 5% scenario captured fewer late orders, while the 20% scenario required twice the workload and produced lower precision and lift. The 10% scenario provides a practical starting balance rather than a proven optimal capacity.  
  
* **Implementation boundary:** Risk ranking should support human review rather than trigger automated operational decisions. The pilot should remain limited until live performance and operational value are validated.

### 5.2 Define monitoring and live validation

* **Pilot monitoring:** Record the number of eligible, queued, and completed reviews in each review cycle to confirm that the process remains within the selected capacity.  
  
* **Ranking performance:** After delivery outcomes become available, compare observed review precision, late-order recall, and lift with the retrospective simulation results.  
  
* **Operational value:** Record the actions taken after review and assess whether reviewed orders received useful and timely operational attention.  
  
* **Live validation:** Evaluate performance across multiple review cycles before deciding whether the ranking remains reliable on new orders.  
  
* **Adjustment rule:** Retain, reduce, or expand the 10% capacity only after considering review workload, ranking performance, and operational value together.  
  
* **Decision separation:** Evaluate whether the model identifies higher-risk orders separately from whether the review process helps improve delivery outcomes.  

### 5.3 Carry forward the main limitations

* **Historical-data limitation:** The findings describe the available historical period and may not represent current or future operating conditions.  
  
* **Feature limitation:** The ranking uses selected order-creation attributes and does not include every operational factor that may influence delivery performance.  
  
* **Association limitation:** Shipping Mode and the other analyzed attributes show associations with late delivery but do not establish causal relationships.  
  
* **Model limitation:** Logistic regression provides a transparent benchmark ranking and was evaluated on one temporal holdout period; it is not confirmed as the best or permanently stable model.  
  
* **Simulation limitation:** The capacity results are retrospective and assume that orders are reviewed from the fixed ranking without operational constraints beyond the selected capacity.  
  
* **Action limitation:** The analysis estimates which orders should be reviewed first, but it does not demonstrate that reviewing an order will prevent late delivery.
