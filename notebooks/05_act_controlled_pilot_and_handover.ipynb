# 05 Act - Controlled Pilot and Final Handover

## 1. Define the Act scope and responsibility boundary

* **Purpose:** Convert the validated Share-stage recommendation into a limited action plan and complete the project handover.  
  
* **Process logic:** Validated findings → controlled pilot → monitored implementation → review decision → final handover.  
  
* **Analyst responsibility:** Define the evidence-based action, implementation measures, evaluation rules, limitations, and handover materials.  
  
* **Business responsibility:** Decide whether to approve the pilot, assign operational owners, conduct reviews, and choose any intervention after human assessment.  
  
* **Scope boundary:** This portfolio project proposes the action and evaluation framework but does not claim that the pilot was implemented or that delivery performance improved.

## 2. Translate the recommendation into a controlled pilot

### 2.1 Confirm the pilot action

* **Purpose:** Define one practical starting action from the AQ1–AQ4 evidence.  
  
* **Action:** Run a controlled pilot that reviews the top 10% of eligible orders ranked by predicted late-delivery risk.  
  
* **Selection rule:** Rank eligible orders from highest to lowest risk score and stop when the 10% review-capacity limit is reached.  
  
* **Evidence basis:** In the held-out-period simulation, the top-10% scenario reviewed **1,353 of 13,525 orders**, captured **17.37%** of late orders, achieved **100.00% review precision**, and produced **1.74x lift**.  
  
* **Interpretation:** The 10% capacity is a practical pilot starting point, not a proven optimal or permanent operating level.

### 2.2 Define the operating workflow and owners

* **Purpose:** Translate the recommendation into a simple repeatable review cycle.  
  
* **Data preparation:** The analyst or reporting owner prepares the latest eligible-order data and applies the validated scoring process.  
  
* **Queue creation:** The reporting owner ranks eligible orders and produces the top-10% review queue.  
  
* **Operational review:** The delivery or operations team reviews queued orders, records the review outcome, and decides whether any operational follow-up is appropriate.  
  
* **Outcome completion:** Delivery outcomes are added after they become available so that ranking and operational results can be evaluated.  
  
* **Initial cadence:** Run the cycle weekly during the pilot, subject to business approval and data availability.

### 2.3 Keep human decision control

* **Purpose:** Prevent the ranking from being treated as an automated operational decision.  
  
* **Decision-support rule:** Risk scores determine review priority only; they do not determine whether an order will be late.  
  
* **Human review rule:** Operational staff assess each queued order before selecting any intervention.  
  
* **No-action claim:** The historical analysis does not show that reviewing or intervening in a queued order will prevent late delivery.  
  
* **Pilot control:** Keep the pilot limited until live ranking performance, workload, and operational usefulness are assessed together.

## 3. Define monitoring and evaluation

### 3.1 Monitor implementation

* **Purpose:** Confirm that the pilot is being operated consistently and within the selected capacity.  
  
* **Volume measures:** Record eligible orders, queued orders, completed reviews, and the percentage of eligible orders reviewed in each cycle.  
  
* **Process measures:** Record review completion rate, review timing, and the operational action taken after review.  
  
* **Capacity check:** Confirm that the queue follows the highest-risk-first rule and remains within the approved capacity.  
  
* **Data-quality check:** Record missing fields, scoring failures, and any changes to the eligible-order definition.

### 3.2 Evaluate ranking and operational outcomes

* **Purpose:** Separate model-ranking performance from the effect of the operational review process.  
  
* **Ranking measures:** Calculate observed review precision, late-order recall, and lift after delivery outcomes become available.  
  
* **Delivery measure:** Track the overall late-delivery rate and the rate among reviewed orders as descriptive context.  
  
* **Operational value:** Assess whether reviews were timely, produced useful follow-up, and were manageable within team capacity.  
  
* **Evaluation boundary:** A strong ranking result does not by itself prove that the review process improved delivery outcomes.

### 3.3 Set the pilot review checkpoint

* **Purpose:** Define how evidence from the pilot should support the next business decision.  
  
* **Review point:** Evaluate results after multiple complete review cycles rather than drawing conclusions from one week.  
  
* **Retain:** Continue the 10% capacity if ranking performance, workload, and operational usefulness remain acceptable.  
  
* **Adjust:** Reduce or expand capacity only after considering review workload, late-order capture, precision, and operational value together.  
  
* **Pause:** Stop and investigate if data quality changes, scoring fails, or live performance differs materially from the historical simulation.

## 4. Record limitations and next steps

### 4.1 Carry forward the main limitations

* **Purpose:** Ensure that the action plan remains within the limits of the available evidence.  
  
* **Historical scope:** Results may not represent current or future operating conditions.  
  
* **Feature scope:** The ranking uses selected order-creation attributes and does not include every operational factor.  
  
* **Association scope:** Shipping Mode and other analyzed attributes show associations but do not establish causation.  
  
* **Model scope:** Logistic regression was evaluated on one temporal holdout period and is not confirmed as the best or permanently stable model.  
  
* **Simulation scope:** Capacity results are retrospective and do not demonstrate the effect of real review activity.

### 4.2 Define proportionate next steps

* **Purpose:** Identify only the improvements justified after live evidence becomes available.  
  
* **First priority:** Validate the existing ranking and review process prospectively before adding modelling complexity.  
  
* **Data update:** Recheck field definitions, missingness, and category changes when new operational data are introduced.  
  
* **Performance update:** Re-evaluate ranking metrics over time and investigate material decline.  
  
* **Future extension:** Consider additional features, alternative models, or formal impact evaluation only if the pilot identifies a clear business need.
