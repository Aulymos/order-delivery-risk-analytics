## 1. Define the business problem and scope

**Purpose and workflow**

- **Purpose:** Turn the broad delivery topic into a clear and limited portfolio problem.
- **Logic:** Business problem → project objectives → portfolio boundary.

**State the problem, objectives, and portfolio boundary**

**Business problem**

**How can historical order data be used to understand late-delivery patterns, identify orders that may require early attention, and prioritise limited review capacity?**

**Project objectives**

- Describe historical delivery performance and selected operational patterns.
- Compare order-creation attributes associated with late delivery.
- Rank later held-out orders by predicted late-delivery risk.
- Compare review capacity at 5%, 10%, and 20%.

## 2. Identify decision users and decisions

**Purpose and workflow**

- **Purpose:** Connect the business problem to the people who may use the analysis and the decisions it supports.
- **Logic:** Assumed users → decision needs → D1–D3.

**Define the assumed users and D1–D3**

The users below are portfolio assumptions rather than verified organisational stakeholders.

| Assumed user | Decision need |
|---|---|
| Operations or fulfilment manager | Understand delivery patterns and allocate limited review capacity. |
| Customer service or account lead | Use a focused order queue for possible follow-up. |

**Decisions**

- **D1 — Historical understanding:** Which delivery patterns, segments, or order profiles require further investigation?
- **D2 — Risk identification:** Which eligible orders in a later period should receive higher review priority?
- **D3 — Limited-capacity prioritisation:** How should fixed 5%, 10%, or 20% review capacity be allocated?

The project ranks and compares orders. It does not automate customer contact or operational action.

## 3. Define AQ1–AQ4

**Purpose and workflow**

- **Purpose:** Convert D1–D3 into four measurable questions that match the Analyze notebook.
- **Logic:** Decision needs → AQ1–AQ4 → planned evidence.

**Define the analytical questions**

- **AQ1 — Delivery performance:** Over the validated historical period, what are the order-level late-order count and late-delivery rate, and how do they change by week or month and across selected operational segments?
- **AQ2 — Associated factors:** Which attributes known at order creation are associated with higher or lower late-delivery rates, after accounting for exposure counts and without making causal claims?
- **AQ3 — Risk ranking:** Using only information available at order creation, can a time-aware model rank orders in a later held-out test period by late-delivery risk more effectively than a non-informative baseline?
- **AQ4 — Prioritisation:** When review capacity is limited to 5%, 10%, or 20% of eligible orders, which orders should be reviewed first, and what late-order recall, precision, and lift would the queue achieve in the held-out test period?

**Question dependencies**

- AQ1 and AQ2 use the validated historical Process outputs.
- AQ3 creates the held-out order-level risk ranking.
- AQ4 reuses the fixed AQ3 ranking and does not fit another model.

**Confirm that the questions are practical and measurable**

- **Specific:** Each question identifies the subject and unit of analysis.
- **Measurable:** Each question uses defined counts, rates, ranking measures, or capacity measures.
- **Action-oriented:** Each question supports D1, D2, or D3.
- **Relevant:** All four questions connect directly to the business problem.
- **Time-bound:** AQ1–AQ2 use the historical period; AQ3–AQ4 use a later held-out period.
- **Feasibility:** Notebook 01 will confirm whether the source can support the questions.

## 4. Set analysis rules and project scope

**Purpose and workflow**

- **Purpose:** Set the few rules needed to keep Prepare, Process, and Analyze consistent.
- **Logic:** Order grain → information boundary → time split → capacity rule → scope.

**Fix the order, timing, and capacity rules**

**Order and target**

- One row represents one unique eligible customer order.
- The target is a binary order-level late-delivery outcome.
- Repeated raw order-item rows must be reconciled before analysis.

**Information boundary**

- Use only fields treated as available at order creation as predictors.
- Retain Order Id for traceability and tie-breaking, not as a predictor.
- Exclude post-order shipping and delivery information from predictive inputs.

**Time and validation**

- Complete weeks run from Monday to Sunday.
- Use the earliest 80% of complete weeks for training and the latest 20% for testing.
- Keep all orders from the same week in the same partition.

**Capacity and ranking**

- Compare 5%, 10%, and 20% of eligible held-out orders.
- Round each review limit up to a whole order.
- Rank orders from highest to lowest predicted risk.
- Use Order Id only as a deterministic tie-breaker.
- Select a preferred capacity only after comparing the scenario evidence.

**Separate in-scope and out-of-scope work**

| In scope | Out of scope |
|---|---|
| Historical KPIs, trends, segments, and selected order profiles | Causal explanations or intervention effects |
| One dummy baseline and one logistic-regression model | Broad model competition or complex tuning |
| One later held-out evaluation period | Production deployment or live monitoring |
| Review simulations at 5%, 10%, and 20% | Automated action or staffing policy |
| Selected final tables and figures | Savings, financial impact, or ROI |
| Historical complete-week delivery trends | Future workload forecasting |
| Portfolio decision-support workflow | Claims about a real organisation's performance |

## 5. Define the evaluation measures

**Purpose and workflow**

- **Purpose:** Define how AQ1–AQ4 will be assessed before results are calculated.
- **Logic:** Analytical question → measure → comparison rule → interpretation.

**Define the minimum measures for AQ1–AQ4**

| ID | Question | Measure | Simple definition |
|---|---|---|---|
| M01 | AQ1–AQ2 | Eligible order count | Number of unique in-scope orders. |
| M02 | AQ1–AQ2 | Late order count | Number of eligible orders labelled late. |
| M03 | AQ1–AQ2 | Late-delivery rate | Late orders divided by eligible orders. |
| M04 | AQ3 | Average Precision | How strongly late orders are concentrated near the top of the ranking. |
| M05 | AQ3 | ROC-AUC | How often late orders rank above non-late orders. |
| M06 | AQ3 | Brier score | Average error in the predicted probabilities; lower is better. |
| M07 | AQ3 | Calibration check | Comparison between predicted and observed late-delivery risk. |
| M08 | AQ4 | Late-order recall | Selected late orders divided by all late orders. |
| M09 | AQ4 | Review precision | Selected late orders divided by reviewed orders. |
| M10 | AQ4 | Lift over baseline | Review precision divided by the overall late-order rate. |

Counts are shown with rates so that small groups are not interpreted without their exposure.

**Set the evaluation and interpretation rules**

- Compare logistic regression with a dummy classifier that does not use the order features.
- Fit preprocessing and both models using the earlier training period only.
- Evaluate both models on the same later held-out period.
- Use continuous risk scores for ranking rather than a fixed 0.5 threshold.
- Reuse the same AQ3 risk scores for all AQ4 capacity scenarios.
- Report historical associations, predictive ranking, and retrospective simulation only.
